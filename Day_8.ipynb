```python
# 1. Write a program to find the second smallest
num = [10,45,23,87,67,89,34]

def second_smallest(l):
    smallest = l[0]
    second = l[1]

    for i in l:
        if i < smallest:
            second = smallest
            smallest = i

        elif i < second and i != smallest:
            second = i

    print("smallest number :", smallest)
    print("second smallest number :", second)

s = second_smallest(num)
print(s)
```

```python
# 2 . Write a function digit_sum(n) that returns the sum of all digits.

def digit_sum(n):
    add=0
    for i in str(n):
        add = add + int(i)
    return add
s=digit_sum(12345)
print(s)
```

```python
# 4 .Write a function find_min(numbers) that returns the smallest number without using min().
l = [4 , 7 , 100 , 2 , 9 , 4 , 56 , 1 , 5]

def find_min(numbers):
    minimum = numbers[0]
    for i in numbers:
        if minimum > i:
            minimum = i
    return minimum
s=find_min(l)
print(s)
```

```python
# 7 .  Write a function common_elements(a, b) that returns the elements present in both lists.

l1 = [1,2,3,4] 
l2 = [3,4,5,6]

def common_elements(a, b):
    common = []

    for i in a:
        for j in b:
            if i == j:
                common.append(i)
    return common 
common_elements(l1,l2)
```

```python
# 9. Write a function that checks whether a number is an Armstrong number.
def Armstrong_number(n):
    original = n
    power = len(str(n))
    total = 0

    while n > 0:
        digit = n % 10
        total = total + digit ** power
        n = n // 10

    if total == original:
        return True
    else:
        return False

print(Armstrong_number(int(input("Enter a number to check number is armstrong number or not : "))))
```
