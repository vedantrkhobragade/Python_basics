```python
# program for area of circle
radius = float(input("Enter the radius of circle:"))
area = 3.14 *(radius**2)
print(f"Area of circle is {area}")
```

```python
# for area of rectangle
height=float(input("Enter the height value:"))
width=float(input("Enter the width value:"))
area= height*width
print(f"Area of rectangle {area}")
```

```python
# for area of square
square=float(input("Enter the value:"))
area=square**2
print(f"Area of square {area}")
```

```python
#for area of triangle

height=float(input("Enter the value:"))
Base=float(input("Enter the vale:"))
area= 1/2* Base*height
print(f"Area of triangle {area}")
           
```

```python
# for Square root
a=int(input("Enter a number:"))
squareroot=int(a**0.5)
print(f"squareroot of {a} is {squareroot}")
```

```python
# for cube root
a=int(input("Enter a number:"))
cuberoot=int(a**(1/3))
print(f"cube root of {a} is {cuberoot}")
```

```python
# Perimeter of a Square
side=float(input("Enter a number:"))
perimeter=4*side
print(f"perimeter of square is {perimeter}")
```

```python
# perimeter of rectangle
length = float(input("Enter the lenght of the rectangle : "))
breadth = float(input("Enter the breadtth of the rectangle : "))
perimeter = 2 * (length + breadth)
print(f"Perimeter of rectangle is {perimeter} ")
```

```python
# Average of three numbers
n1 = int(input("Enter 1st number : "))
n2 = int(input("Enter 2nd number : "))
n3 = int(input("Enter 3rd number : "))

sum = n1 + n2 + n3
average = sum / 3
print(f"The average of {n1},{n2} and {n3} is {average}")
```
