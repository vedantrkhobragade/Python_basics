```python
# Area of a Circle Sector

angle=int(input("Enter the angle:"))
radius=int(input("Enter the radius:"))

area=(angle/360) * 3.14 * (radius**2)
print(area)
```

```python
# Sum of Digits

num=input("Enter a number:")
i=0
for j in num:
    i = i + int(j)
print(f"The sum of all digits is {i}")
```

```python
# Reverse a number 
num=int(input("Enter your number:"))
reverse=0

while num>0:
    digit=num % 10
    reverse =  reverse * 10 + digit 
    num= num// 10
print(reverse)
```

```python
# for odd and even

num = int(input("Enter a number : "))

if num % 2== 0:
    print(f"{num} is even")
else:
    print(f"{num} is odd")
```

```python
# Armstrong Number

num = int(input("Enter a number: "))

original_num = num
digits = len(str(num))
total = 0

while num > 0:
    digit = num % 10
    total = total + digit ** digits
    num = num // 10

if total == original_num:
    print(f"{original_num} is an Armstrong number")
else:
    print(f"{original_num} is not an Armstrong number")
```

```python
# Find GCD

n1 = int(input("Enter the 1st number : "))
n2 = int(input("Enter the 2nd number : "))

while n2 != 0:
    remainder  = n1 % n2
    n1 = n2
    n2 = remainder
print(f"GCD  is {n1}")
```

```python
# Fibonacci Series 
num = int(input("Enter a numeber : "))

first = 0
second = 1

for i in range(num):
    print(first)
    next_number = first + second
    first = second
    second = next_number
```
