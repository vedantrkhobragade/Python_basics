```python
# 1. Write a function that counts how many times each number occurs.
def frequency(n):
    visited=[]

    for i in n:
        if i not in visited:
            count = 0

            for j in n:
                if i == j:
                    count+=1
            print(i,"occurs",count, "times")
            visited.append(i)
frequency([2,4,2,5,5,76,4])
```

```python
# 2. Write a function that takes a list and returns a new list without duplicate values.

numbers = [1, 2, 2, 3, 1, 4, 3]
def without_duplicate(n):
    l = []
    for i in n:
        if i not in l:
            l.append(i)
    return l
s=without_duplicate(numbers)
print(s)
            
```

```python
# 3.  Write a function unique_values(numbers) that returns a list containing only values that occur once.
def unique_values(numbers):
    result = []

    for i in numbers:
        count = 0

        for j in numbers:
            if i == j:
                count += 1

        if count == 1:
            result.append(i)

    return result


u = unique_values([2, 3, 2, 5, 3, 7])
print(u)
```

```python
# 4. Write a function frequency(numbers, target) that returns how many times target occurs.
def frequency(numbers, target):
    count = 0

    for i in numbers:
        if i == target:
            count += 1

    return count


f = frequency([2, 3, 2, 5, 3, 2], 2)
print(f"it occurs {f} times")
```

```python
# 5 .  Write a function find_missing(numbers) where the list contains numbers from 1 to n with one number missing.

def find_missing(numbers):
    n = len(numbers)+1
    for i in range(1,n+1):
        if i not in numbers:
            return i 
n=find_missing([1,2,3,4,6])
print(f"The missing number is {n}")
```

```python
# 6.  Write a function largest_difference(numbers) that finds the difference between the largest and smallest number.
def largest_difference(numbers):
    largest=numbers[0]
    smallest=numbers[0]
    difference=numbers[0]

    for i in numbers:
        if largest < i:
            largest = i
        if smallest > i:
            smallest = i

    difference = largest - smallest
    return difference
n=largest_difference([4,2,1,6,34,7,45,3,12,])
print(f"The largest difference is {n}")
```

```python
# 7. Write a function separate_numbers(numbers) that creates two lists: even numbers , odd numbers
def separate_numbers(numbers):
    Even=[]
    Odd=[]

    for i in numbers:
        if i %2==0:
            Even.append(i)
        else:
            Odd.append(i)
    return Even,Odd
s=separate_numbers([1,2,3,4,5,6,7,8])
print(s)
            
    
```

```python
# 8. Write a function separate_numbers(numbers) that creates two lists: numbers greater than 10 and numbers less than or equal to 10.
def separate_numbers(numbers):
    numbers_greater_10=[]
    numbers_less_equalto=[]

    for i in numbers:
        if i > 10:
            numbers_greater_10.append(i)
        if i <= 10:
            numbers_less_equalto.append(i)
    return numbers_greater_10,numbers_less_equalto
numbers = [5, 12, 8, 15, 10, 20, 3]

greater, smaller = separate_numbers(numbers)

print("Numbers greater than 10:", greater)
print("Numbers less than or equal to 10:", smaller)   
```

```python
# 9. ATM Withdrawal
def atm(amount,balance):
    if amount > balance:
        print("Insufficient balance")
    elif amount % 100!=0:
        print("Amount is not valid")
    elif amount > 20000:
        print("Enter the amount under 20000")
    else:
        balance = balance - amount
        print("Withdrawal successful")
        print(f"Available balance is {balance}")
amount=int(input("Enter Withdrawal amount"))
atm(amount,10000)
```
