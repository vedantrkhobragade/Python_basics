```python
# Percentage of marks
Hindi = int(input("Enter your marks in Hindi: "))
English = int(input("Enter your marks in English: "))
Marathi = int(input("Enter your marks in Marathi: "))
Social = int(input("Enter your marks in Social: "))
Science = int(input("Enter your marks in Science: "))

total = Hindi + English + Marathi + Social + Science

print(f"The total of all subjects is {total} out of 500")

percentage = (total / 500) * 100

print(f"{percentage}%")

if percentage >= 31 and percentage < 45:
    print("Passed with C grade")
elif percentage >= 45 and percentage < 60:
    print("Passed with B grade")
elif percentage >= 60 and percentage < 75:
    print("Passed with B+ grade")
elif percentage >= 75 and percentage < 85:
    print("Passed with A grade")
elif percentage >= 85 and percentage <= 100:
    print("Passed with A+ grade")
else:
    print("Failed")

```
