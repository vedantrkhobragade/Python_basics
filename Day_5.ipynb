```python
# 1 . Reverse a number
num=int(input("Enter a number:"))
reverse = 0

while num>0:
    digit = num %10
    reverse = reverse * 10 + digit
    num = num//10
print(reverse)
```

```python
# 2. Count digits
num=int(input("Enter your number:"))
count=0

while num>0:
    num=num//10
    count +=1
print("number of digits",count)
        
```

```python
# 3 . Palindrome number
num=int(input("Enter a number:"))

original=num
i=1
reverse=0

while num>0:
    digit=num%10
    reverse=reverse * 10 + digit
    num=num//10
if original==reverse:
    print("It is a palindrome number.")
else:
    print("It is not a palindrome number.")
```

```python
# 4 . Keep asking until correct
answer=True
while answer:
    num=int(input("Enter a number:"))
    if 1<=num<=10:
        print(f"The {num} is in range")
        answer=False
```

```python
# 5 . Find the second largest number
numbers=[10, 25, 8, 45, 32, 18]
largest=numbers[0]
second=numbers[0]

for i in numbers:
    if largest < i:
        second = largest
        largest=i
    elif i > second and i !=largest:
        second=i
print(f"{second} is the second largest ")
```

```python
# 6 . Count vowels
word=input("Enter your word:")
vowels=['a','e','i','o','u']
count=0
for i in word:
    for j in vowels:
        if i==j:
            count = count + 1
print(f" the vowels in this word are {count}")
```

```python
# 7 . Find common element
a=[1,2,5,7,9]
b=[1,3,5,8,9]

print("The common numbers in two list are",end=' ')

for i in a:
    for j in b:
        if i==j:
            print(i,end=' ')
```

```python
# 8. Find duplicate numbers
numbers=[2, 5, 2, 8, 5, 9, 2]
duplicate=[]

for i in numbers:
    for j in numbers:
        if i==j and i not in duplicate:
            duplicate.append(i)
print(duplicate)
```
