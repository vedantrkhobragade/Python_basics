```python
# 1 . Write a function digit_product(n) that multiplies all digits
def digit_product(n):
    product = 1
    while n > 0:
        digit= n % 10
        product = product * digit
        n = n//10
    return product
s=digit_product(1234)
print(s)
```

```python
# 2. Write a function largest_digit(n) without converting the number to a string.
def largest_digit(n):
    largest = 0

    while n > 0:
        digit = n % 10

        if digit > largest:
            largest=digit
        n = n//10
    return largest 
s=largest_digit(76582)
print(s)
```

```python
# 3. Write a function count_even_digits(n). 
def count_even_digits(n):
    count=0

    while n>0:
        digit = n % 10

        if digit % 2==0:
            count +=1

        n = n//10
    return count
s=count_even_digits(12678)
print(s)
```

```python
# 4 . Write a function smallest_digit(n) without converting the number to a string.
def smallest_digit(n):
    smallest = n % 10

    while n > 0:
        digit = n % 10

        if digit < smallest:
            smallest = digit
        n = n//10
    return smallest
s=smallest_digit(48923)
print(s)
```

```python
# 5. Write a function that returns the first repeated number in a list.

def repeated_number(n):

    seen = []
    i = 0

    while i < len(n):

        if n[i] in seen:
            return n[i]

        seen.append(n[i])
        i += 1

    return None

r = repeated_number([1,3,7,1,4,2,5,2])
print(r)
```

```python
# 7. Write a function largest_digit(n) that returns the largest digit in a number using only a while loop.
def largest_digit(n):
    largest = 0

    while n > 0:
        digit = n % 10
        if digit > largest:
            largest = digit

        n = n//10
    return largest
print(largest_digit(58321)) 
```

```python
# 8. Write a function product_of_digits(n) using only a while loop.

def product_of_digits(n):
    product = 1
    while n > 0:
        digit = n % 10
        product = product * digit
        n = n//10
    return product

p = product_of_digits(1235)
print(p)
```

```python
# 9.Write a function count_vowels(text) that counts the vowels in a string.
def count_vowels(text):
    count = 0

    for char in text.lower():
        if char in "aeiou":
            count +=1
    return count
print(count_vowels("vedant"))
```

```python
# 10.Write a function common_elements(list1, list2) that returns the elements present in both lists, without duplicates in the result.
def common_elements(list1, list2):
    common = []

    for num in list1:
        if num in list2 and num not in common:
            common.append(num)
    return common
print(common_elements([1, 2, 3, 4], [3, 4, 5, 6]))
```
