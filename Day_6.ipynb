```python
# 1 .WAP to create list of prime number from 1 to 100

Prime_number = []

for num in range(2,101):
    count = 0
    for i in range(1,num + 1):
        if num % i == 0:
            count += 1

    if count == 2:
        Prime_number.append(num)
print(Prime_number)
```

```python
# 4  .Separte positive and nagative
numbers=[1,4,3,5,6,-1,-2,-3,-6,0]

positive=[]
negative=[]
zero=[]

for i in numbers:
    if i>0:
        positive.append(i)
    elif i<0:
        negative.append(i)
    elif i==0:
        zero.append(i)
print(f"positive list is {positive}")
print(f"negative list is {negative}")
print(f"zero is {zero}")

        
```

```python
# 5 .Find Duplicate Numbers
numbers=[4, 7, 2, 4, 9, 7, 5, 2, 8]

for i in range(len(numbers)):
    for j in range(i+1,len(numbers)):
        if numbers[i]==numbers[j]:     
            print(numbers[i])
            break
```

```python
# 7. Create a new list for odd and even
num = [3,2,4,5,9,7,1,6,11,13]
print(f"Original list = {num}")
Even_list=[]
Odd_list=[]
for i in num:
    if i %2==0:
        Even_list.append(i)
    else:
        Odd_list.append(i)
print(f"Even_list is {Even_list}")
print(f"Odd_list is {Odd_list}")
```
