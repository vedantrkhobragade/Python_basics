```python
#  1 .Shopping discount 
shopping_amount=int(input("Enter your shopping amount:"))

shopping_20=shopping_amount * (20/100)
shopping_15=shopping_amount * (15/100)
shopping_10=shopping_amount * (10/100)

if shopping_amount>=3000:
    print(f"you get 20% discount and your total bill is {shopping_20}")
elif shopping_amount >= 1500 and shopping_amount <3000:
    print(f"you get 15% discount and your total bill is {shopping_15}")
elif shopping_amount >=1000 and shopping_amount < 1500:
    print(f"you get 10% discount and your total bill is {shopping_10}")
else:
    print("shopping_amount")
```

```python
# 2 . ATM MACHINE
balance_amount=float(input("Enter your balance amount:"))
withdraw_amount=float(input("Enter withdraw amount:"))

if withdraw_amount > balance_amount:
    print("insufficient balance")
elif withdraw_amount %100 !=0:
    print("Invalid amount , try again with a amount which is multiple of 100")
elif balance_amount - withdraw_amount <=500:
    print("Balance amount should remain atleast 500")
else:
      balance_amount -= withdraw_amount
      print("Transcation succesfull")
      print("Remaining amount : ",balance_amount)
```

```python
# 3. Three Numbers – Middle Number 
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
c=int(input("Enter third  number:"))

if b > a > c and c > a > b:
    print(f"{a} is a middle number")
elif a > b > c and c > b > a :
    print(f"{b} is a middle number")
else :
    print(f"{c} is a middle number")
```
