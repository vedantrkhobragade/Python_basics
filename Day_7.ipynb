```python
# 4 . Function for prime number

def prime(n1):
    for i in range(2,n1):
        if n1 % i ==0:
            return False
    else:
            return True
prime(7)
```

```python
#  6. Write a function count_vowels(word) that takes a string and returns the number of vowels.
def  count_vowels(word):
    count =0
    vowels=['a','e','i','o','u']

    for i in word:
        if i in vowels:
           count +=1
    
    return count

s=count_vowels("vedant")
print(s)
```

```python
#  7 .Write a function factorial(n) that returns the factorial of a number.

num=int(input("Enter your number:"))

def fact(n):
    factorial=1
    for i in range(1,n+1):
        factorial = factorial * i
    return factorial
f=fact(num)
print(f" Factorial of {num} is {f}")
```

```python
# 8. Write a function count_digits(n) that takes an integer and returns the number of digits.
def count_digits(n):
    count=0
    for i in str(n):
        count+=1

    return count
print(count_digits(456))
```

```python
# 9. Write a function second_largest(numbers) that takes a list and returns the second-largest number.
l = [2,3,5,10,90,89,8]

def second_largest(numbers):

    largest = l[0]
    second = l[0]

    for i in numbers:
        if i > largest:
            second = largest
            largest = i
        elif i > second and i != largest:
            second = i
    return second

s = second_largest(l)
print(f"The second largest number from list is {s}")  
```

```python
# 10 . Write a function reverse_number(n) that takes a number and returns its reverse.
def reverse_number(n):
    reverse = 0

    for i in range(len(str(n))):
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10

    return reverse

print(reverse_number(56789))
```
